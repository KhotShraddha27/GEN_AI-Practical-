```python
from openai import OpenAI
import os
client =OpenAI(api_key="xxxx")
```

```python
def call_llm(messages,model="gpt-4o-mini",temprature=0.7):
  response = client.chat.completions.create(
      model=model,
      messages=messages,
      temperature=temprature
  )
  return response.choices[0].message.content
```

```python
#Zero-Shot prompting
messages = [
    {
        "role":"user",
        "content":"Explain quantum computing in simple terms."
    }
]
print(call_llm(messages))
```

```python
#One-Shot prompting
messages = [
    {
        "role":"user",
        "content":
         (
           "Summarize the following text in 3 bullet points .\n\n"
            "Text: Large Language Models are neural networks trained on vast datasets...")
    }
]
print(call_llm(messages))
```

```python
# Few-Shot prompting: Explanation style learning
messages = [
    {
        "role": "user",
        "content":
        """
        Explain AI simply:\nAI is like teaching machines to think and learn.
        Explain Machine Learning simply:\nML allows machines to learn from data without being explicitly programmed.
        Explain Large Language Models simply:"
        """
    }
]

print(call_llm(messages))
```

```python
# Instruction-Based Prompting: Explanation
messages = [
    {
        "role": "user",
        "content": (
            "Explain the concept of Generative AI in simple terms."
            "Limit your answer to 4 lines  with bullets."
        )
    }
]

print(call_llm(messages))
```

```python
# Instruction-Based Prompting: Reasoning
messages = [
    {
        "role": "user",
        "content": (
            "Explain step-by-step how a Large Language Model generates an answer, in short ."
        )
    }
]

print(call_llm(messages))
```

```python
# Role-Based Prompting
messages = [
    {
        "role": "user",
        "content": (
            "You are a computer science professor. "
            "Explain Large Language Models to a first-year engineering student."
        )
    }
]

print(call_llm(messages))
```
